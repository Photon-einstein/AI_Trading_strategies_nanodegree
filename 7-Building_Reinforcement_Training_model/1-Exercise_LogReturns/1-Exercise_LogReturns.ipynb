# 1. Raw vs Log Returns

**Objective:** Calculate the raw profit and log returns for a given set of stock prices, and demonstrate the additive and symmetric properties of these.

---

## Instructions

Assume you have the following daily closing prices of a stock over a week: `[100, 105, 103, 107, 106]`.

### Part 1

Calculate the **raw profit**, **raw returns**, and **log returns** for each day relative to the previous day.

| Metric | Formula |
|---|---|
| Raw Profit | `sell_price - buy_price` |
| Raw Return | `sell_price / buy_price` |
| Log Return | `ln(sell_price / buy_price)` = `ln(sell_price) - ln(buy_price)` |

### Part 2

Perform the following tasks in order:

**a.** Calculate the raw profit, raw returns, and log returns for the **entire period** (i.e. buy at `t=0` and sell at `t=4`).

**b.** Demonstrate and explain how:

- **i.** The full-period **raw profit** relates to the *additive* and *symmetrical* properties of the individual raw profits from Part 1.
- **ii.** The full-period **log return** relates to the *additive* and *symmetrical* properties of the individual log returns from Part 1.
- **iii.** The full-period **raw return** relates to the *lack* of additive property of the individual raw returns from Part 1.

```python
import pandas as pd
import numpy as np
```

```python
data = {"Close": [100, 105, 103, 107, 106]}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.index.name = "Time"
df
```

```python
df["Daily Returns"] = df["Close"].diff()
df
```

```python
df["Daily Log Returns"] = np.log(df["Close"] / df["Close"].shift(1))
df
```

```python
# Part 1 — Raw Return (ratio): sell_price / buy_price
df["Raw Return"] = df["Close"] / df["Close"].shift(1)
df
```

```python
# Part 2a — Full-period metrics (buy at t=0, sell at t=4)
buy_price = df["Close"].iloc[0]  # 100
sell_price = df["Close"].iloc[-1]  # 106

full_raw_profit = sell_price - buy_price
full_raw_return = sell_price / buy_price
full_log_return = np.log(sell_price / buy_price)

print(f"Buy price  (t=0): {buy_price}")
print(f"Sell price (t=4): {sell_price}")
print(f"\nFull-period Raw Profit : {full_raw_profit:.4f}")
print(f"Full-period Raw Return : {full_raw_return:.4f}")
print(f"Full-period Log Return : {full_log_return:.4f}")
```

```python
# Part 2b — Demonstrate additive and symmetric properties

# ── i. Raw Profit ─────────────────────────────────────────────────────────────
sum_raw_profit = df["Daily Returns"].dropna().sum()
print("=== i. Raw Profit ===")
print(f"Full-period raw profit       : {full_raw_profit:.4f}")
print(f"Sum of daily raw profits     : {sum_raw_profit:.4f}  → additive ✓")
reverse_raw_profit = buy_price - sell_price
print(
    f"Forward + reverse profit     : {full_raw_profit + reverse_raw_profit:.4f}  → symmetric ✓"
)
print()

# ── ii. Log Return ────────────────────────────────────────────────────────────
sum_log_return = df["Daily Log Returns"].dropna().sum()
reverse_log_return = np.log(buy_price / sell_price)
print("=== ii. Log Return ===")
print(f"Full-period log return       : {full_log_return:.4f}")
print(f"Sum of daily log returns     : {sum_log_return:.4f}  → additive ✓")
print(
    f"Forward + reverse log return : {full_log_return + reverse_log_return:.4f}  → symmetric ✓"
)
print()

# ── iii. Raw Return ───────────────────────────────────────────────────────────
sum_raw_return = df["Raw Return"].dropna().sum()
product_raw_return = df["Raw Return"].dropna().prod()
print("=== iii. Raw Return (NOT additive) ===")
print(f"Full-period raw return       : {full_raw_return:.4f}")
print(
    f"Sum of daily raw returns     : {sum_raw_return:.4f}  → NOT equal to full-period ✗"
)
print(
    f"Product of daily raw returns : {product_raw_return:.4f}  → multiplicative, equals full-period ✓"
)
```
